```python
l=[10,20,100,40]
# [40,100,20,10]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        l[i],l[j]=l[j],l[i]
print(l)
```

```python
# Sorting:(logic)
# Ascending
l=[10,7,100,40]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        if l[i]>l[j]:
            l[i],l[j]=l[j],l[i]
print(l)
```

```python
# Descending
l=[10,7,100,40]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        if l[i]<l[j]:
            l[i],l[j]=l[j],l[i]
print(l)
```
