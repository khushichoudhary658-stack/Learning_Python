```python
sub1=int(input("enter marks of sub1: "))
sub2=int(input("enter marks of sub2: "))
sub3=int(input("enter marks of sub3: "))
sub4=int(input("enter marks of sub4: "))
sub5=int(input("enter marks of sub5: "))
obt_marks= sub1+sub2+sub3+sub4+sub5
per=(obt_marks/500)*100
print("percentage:",per,"%")
if per>=65:
    print("first division")
elif per>=50:
    print("second division")
elif per>=33:
    print("third division")
else:
    print("fail")
```

```python
# WAP to check wheather a given traingle is valid or invalid on the basis of side condition?
# if triangle is valid then also check is it equilateral, scalen, or isoceles?

s1= int(input("enter side1"))
s2= int(input("enter side2"))
s3= int(input("enter side3"))
if s1+s2>s3 and s2+s3>s1 and s1+s3>s2:
    if s1==s2 and s2==s3:
        print("equilateral triangle")
    elif s1==s2 or s2==s3 or s3==s1:
        print("isoceles triangle")
    else:
        print("scalen triangle")
else:
    print("invalid triangle")
```

```python
# WAP to check wheather a given year is leap year or not?
year=int(input("enter any number: "))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"leap year")  # century
        else:
            print(year, "not a leap year") # century
    else:
        print(year, "leap year")  # non-century
else:
    print(year,"not a leap year") #non-century
```

```python
# Eligibility for special programs
# Objextive: WAP that determines wheather a student is eligible for an honors program or need remedial classes bases on their performance in diff sub.
# Details: Eligibility Criteria:
# Honors: Average score of 85 or above across all sub and no sub below 70.
# Remedial: Any sub below 50.
# INPUT: Scores in three subjects: Math, Science, and Language (out of 100)
# OUTPUT: "Eligible for Honors", "Needs Remedial", "Standard Placement".

math=int(input("enter math's marks: "))
science=int(input("enter science marks: "))
language=int(input("enter language marks: "))
avg=(math + science + language)/3
if avg >= 85 and math >=70 and science >=70 and language >=70:
    print("eligible for honors")
elif math < 50 or science < 50 or language < 50:
    print("need remedial")
else: 
    print("standard placement")
```
