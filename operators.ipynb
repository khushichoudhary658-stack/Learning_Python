# operators

### Arithematic Operator +, -, *, /, //, %, **

```python
a=10
b=2
print("Addition:-",a+b)
print("Subtraction:-",a-b)
print("Multiply:-",a*b)
print("Simple Division:-",a/b)
print("Floor Division:-",a//b)
print("Modulus:-",a%b)
print("Exponential Power:-",a**b)
```

```python
# Assignment Operater:- =, +=, -=, /=, *=,....
a=10
b=5
c=a+b
a+=10     # a=a+10
b-=2      # b=b-2
c*=5      # c=c*5
a/=b      # a=a/b
c=a+b
print(a,b,c)
```

```python
# Comparosion Operator:- ==, !=, >=, <=, >, <
a=10
b=8
print(a==b)
print(a!=b)
print(a>=b)     # a>b or a==b
print(a<=b)
print(a>b)
print(a<b)
```

```python
# Logical Operator:- and, or not
# id():- It is predefined function that return memory address of given object.

a=10
b=8
c=12

print(a>b and b>c)
print(a>b or b>c)
print(a>b and not b>c)
```

```python
# Identity Operator:- is, is not

a=10
b=8
print(id(a))
print(id(b))
print(a is b)
print(a is not b)
```

```python
c=20
d=20
print(id(c))
print(id(d))
print(c is d)
print(c is not d)
```

```python
# Membership Operator:- in, not in

a="hello"

print("h" in a)
print("h" not in a)
print("H" in a)
```

```python
# Bitwise Operator: &, |,^

a=10
b=8
print(a&b)
print(a|b)
print(a^b)
```

```python
# bin():- it is predefined function always returns binary of given integer object.

a=10
b=8
print(bin(a))
print(bin(b))
print("------")
print(bin(a&b))
print(a&b)
print("_*_*_*_*_")
print("_*_*_*_*_")
print(bin(a))
print(bin(b))
print("------")
print(bin(a|b))
print(a|b)
print("_*_*_*_*_")
print("_*_*_*_*_")
print(bin(a))
print(bin(b))
print("------")
print(bin(a^b))
print(a^b)
```
