```python
# WAP to print sum of first n natural number?
n=int(input("enter any number: "))
sum=0
for i in range(1,n+1):
    sum=sum+i
print(sum)
```

```python
# WAP to print sum of first n even number and first n odd number?
n=int(input("enter any number: "))
sum_even=0
sum_odd=0
for i in range(1,(2*n)+1):
    if i%2==0:
        sum_even=sum_even+i
    else:
        sum_odd=sum_odd+i
print("sum of first",n,"even number",sum_even)
print("sum of first",n,"odd number",sum_odd)
```

```python
# WAP to print factorial of n number?   : [f(n)=f(n-1)]
n=int(input("enter any number: "))
fact=1
for i in range (1,n+1):
    fact=fact*i
print(fact)
```

```python
# Fibonacci Series: It is the sum of the previous 2 terms   :{f(n)= f(n-1)+f(n-2)}

# WAP to print Fibinacci series upto nth term? take n as a user input at runtime.
a=0
b=1
n=int(input("How many term: "))
print(a,b,end=" ")
for i in range(2,n):
    c=a+b
    print(c,end=" ")
    a=b
    b=c
```

```python
# WAP to check wheather given number is prime number or not?

n=int(input("enter any number: "))
if n==0 or n==1:
    print(n,"not a prime")
elif n>1:
    for i in range(2,n):
        if n%i==0:
            print(n,"not a prime")
            break
    else: 
        print(n,"prime number")
else: 
    print(n,"not a prime")
```

```python
# WAP to print prime number between start and end take start and end as a user input at rumtime?

start= int(input("enter any number: "))
end= int(input("enter any number: "))
for n in range(start, end+1):
    if n>1:
        for i in range(2,n):
            if n%i==0:
                break
        else:
            print(n,end=" ")
```
