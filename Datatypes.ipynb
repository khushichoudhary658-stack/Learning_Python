```python
# Data Type:- Data Type in python define the type of value stored in a variable.

# On the basis of Mutability:-
# 1. Mutable  2. Immutable
# Mutable Object:- List, Dictionary, Set.
# Immutable Object:- int, float, string, tupple, frozenset.
```

```python
# type():- Type function is a predefined function that returns the data type or class of perticular object.
a=10
print(a)
print(type(a))
```

```python
a=10.5
print(a)
print(type(a))
```

```python
a=10
b=2
print(a,b)
print(type(a),type(b))
c=a/b
print(c)
print(type(c))
```

```python
# input():- it is a per-defined function use to take input from uses at a run time.
# always take string as an input.
a=input("enter any number: ")
b=input("enter any number: ")
c=a+b
print(c)
print(type(c))
```

```python
# typecasting():- it is process of converting a variable(object) from on datatype to another datatype.

a='10'
print(a)
print(type(a))
b=int(a)
print(b)
print(type(b))
```

```python
s=""    #empty string
print(s)
print(type(s))
```

```python
s="hello"
print(s)
print(type(s))
print(s[0])       # indexed
s[0]="z"          #immutable- 'str' object does not support item assignment
print(s)
```

```python
# List:- collection of elements.
# denoted by [] with csv.
# ordered, indexed, mutable and heterogeneous in nature.
l=[]     # empty list
print(l)
print(type(l))
```

```python
l=[10,20,30,10,"hello",10.5]
print(l)
print(type(l))
print(l[-1])    # indexed
l[0]=100        # mutable- it dose support item assignment.
print(l)
```

```python
# tuple():- collection of elements.
# denoted by () with csv.
# ordered,indexed, immutable and heterogeneous in nature.
t=()
print(t)
print(type(t))
```

```python
# single valued tuple
t=(10,)
print(t)
print(type(t))
```

```python
t=(10,20,30,10,"hello",10.5)
print(t)
print(type(t))
print(t[0])   # indexed
t[0]=100     #immutable- 'tuple' object does not support item assignment
print(t)
```

```python
# Dictionary- collection of elements
# denoted by {k:v}, key must be unique but values can be duplicate.
# ordered, unindexed and mutable.
d={}             # empty dictionary
print(d)
print(type(d))
```

```python
d={"name":"abc","subject":"DA","fees":30000}
print(d)
print(type(d))
print(d["name"])   # unindexed
d["name"]="pqr"    # mutable
print(d)
```

```python
d={"name":"abc","subject":"DA","fees":30000}
print(d)
print(type(d))
d["Nmae"]="pqr"
print(d)
```

```python
d={}
print(d)
d["name"]="abc"
print(d)
```

```python
d={"name":["abc","pqr","Da"],"subject":["da","da","de"]}
print(d)
print(d["name"][-1])
d["name"][-1]="xyz"
print(d)
```

```python
# set - collection of unique element.
# denoted by {} with csv.
# unordered, unindexed, mutable and heterogenous in nature.
s=set()      # set() - set constructor
print(s)
print(type(s))
```

```python
s={10,20,10,20,"hello",10.5,10,20,100}
print(s)
print(type(s))
s.add(1000)   # mutable
print(s)
```

```python
# frozenset - collection of unique elements
# denoted by a constructor.
# unordered, immutable, unindexed and heterogeneous in nature.
f=frozenset()
print (type(f))
```

```python
f=frozenset({10,20,10,20,"hello",10.5})
print(f)
print(type(f))
```
