```python
# string - ordered sequence of character and special characters
# denoted by '' or ""
# indexed and immutable
s=''
print(s)
print(type(s))
```

```python
# index - position of a single character in a given string
# first character of any string is at 0 index.
# last character is at -1 index.

# indexing - it is a technique for accessing individual character from a given string.

s="welcome to bhopal"
print(s[0])
print(s[-1])
print(s[7])
print(s[-4])
```

```python
# slicing - it is a technique for extracting substring from a given string.
# strobj[start:stop:step]
# start - by default - 0 index
# stop - by default end of the string  or n-1
# step - by default  - + 1 increment
s="welcome to bhopal"
print(s[::])
print(s[:7:])
print(s[::-1])      # start -> -1 # stop - end of the string --> reverse string
print(s[-1:-7:-1])
```

```python
s="welcome to city of lakes"
print(s[1:15:2])
print(s[-2:-16:-2])
print(s[:11:])
print(s[0:7:2])
print(s[-1:-14:-2])
print(s[1: :3])
```

```python
# iteration 
# direct iteration 
s="hello"
for i in s:
    print(i)
```

```python
# indirect iteration:- with help of range function
# len() - no. of characters in a given string.
s="hello"
for i in range(len(s)):
    print(i,s[i])  
```

```python
# wap to print number of vowels, no. of conconents and the no. of space in a given string.
s="welcome to city of lakes"
c_v=0
c_c=0
c_s=0
for i in s:
    if i in "aeiouAEIOU": 
        c_v=c_v+1
    elif i==' ':
        c_s=c_s+1
    else:
        c_c=c_c+1
print("count of vovel",c_v)
print("count of consonents",c_c)
print("count of spaces", c_s)
```

```python
# wap to print reverse of any given string.
s=input("enter any string: ")
rev=''
for i in s:
    rev=i+rev
print(rev)      # by logic
print(s[::-1])  # by slicing
```

```python
# wap to check weather a given string is palindrome string or not?
s=input("enter any string: ")
rev=''
for i in s:
    rev=i+rev
if s==rev:
    print(s,"palindrome string")
else:
    print(s,"no a palindrome string")
```

```python
# count():- return occurrence of any character in a given string.
# strobj.count(char/substring,[start],[stop])
# start - by default - 0
# stop - by default - end of the string
# stop - n ----> (n-1)
s="welcome to bhopal"
print(s.count('o'))
print(s.count('o',5))
print(s.count(' '))
print(s.count('bhopal'))
print(s.count('z'))
print(s.count('o',5,9))
```

```python
# index:- return position of character/substring from a given string.
# strobj.index(char/substring,[start],[stop])
# always return first occurence.
# index through error if substring is not found
s="welcome to bhopal"
print(s.index("o"))
print(s.index('o',5))
print(s.index('bhopal'))
print(s.index('o',5,9))
```

```python
# find:- return position of character/substring from a given string.
# strobj.find(char/substring,[start],[stop])
# always return first occurrence.
# Find returns -1 if substring is not found
s="welcome to bhopal"
print(s.find("o"))
print(s.find('o',5))
print(s.find('bhopal'))
print(s.find('o',5,9))
```

```python
# split() 
# strobj.split([separator],[maxsplit])
# separator-- space
# maxsplit - max. occurence separator
# always return a list of string.
s="welcome to city of lakes"
l=s.split()
print(l)
```
