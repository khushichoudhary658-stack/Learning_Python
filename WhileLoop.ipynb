```python
# wap to ptrint reverse of any given no.
n=int(input("enter any number: "))
rev=0
while n!=0:
    dig=n%10          #Latsdigit
    rev=rev*10+dig    #output
    n=n//10
print(rev)
```

```python
# wap to print reverse no. between start and end take start and end as a user input at runtime.
start=int(input("enter start number: "))
end= int(input("enetr end number: "))
for n in range(start,end+1):
    rev=0
    while n!=0:
        dig=n%10
        rev=rev*10+dig
        n=n//10
    print(rev)
```

```python
# Find the sum of digits of a number.
n=int(input("enter any number: "))
sm=0
while n!=0:
    dig=n%10
    sm=sm+dig
    n=n//10
print(sm)
```

```python
# Find the product of digits.
n=int(input("enter any number: "))
pro=1
while n!=0:
    dig=n%10
    pro=pro*dig
    n=n//10
print(pro)
```

```python
# Find the largest digit.
n=int(input("enetr any number: "))
max=0
while n!=0:
    dig= n%10
    if dig>max:
        max=dig
    n=n//10
print(max)
```

```python
# Find the smallest digit.
n=int(input("enetr any number: "))
min=9
while n!=0:
    dig=n%10
    if dig<min:
        min=dig
    n=n//10
print(min)
```
