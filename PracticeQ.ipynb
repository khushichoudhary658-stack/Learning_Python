```python
# Check Whether a Year Is a Leap Year Write a Python program that accepts a year from the user and checks whether it is a leap year.
# A year is a leap year if it is divisible by 400, or if it is divisible by 4 but not by 100. 
# INPUT REQUIRED: Input required: A year (integer). 
# EXPECTED OUTPUT: Expected output: Display whether the year is a leap year or not. 
year=int(input("enter any number:"))
if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year, "leap year")
        else:
            print(year, "not a leap year")
    else:
        print(year, "leap year")
else:
    print(year, "not a leap year")
```

```python
# 10. Check Whether a Character Is an Alphabet Write a Python program that accepts a single character and checks whether it is an English alphabet letter. 
# Reject digits, spaces and special characters. Also check that the user entered exactly one character. 
# INPUT REQUIRED: Input required: One character. 
# EXPECTED OUTPUT: Expected output: Display whether the character is an alphabet or not. 

ch= input("enter any character: ")
if len(ch)==1:
    if(ch >='A' and ch <='Z') or (ch >= 'a' and ch <= 'z'):
        print(ch, 'is an alphabet')
    else:
        print(ch, 'is not an alphabet')
else:
    print("enter only one character")
```
